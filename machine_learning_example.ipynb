4.1. Predicting fraudulent financial transactions using a data set where most transactions are labelled as fraudulent or legitimate.

- It is appropriate to utilise semi-supervised machine learning algorithms for this problem because:

  - There is a large amount of labelled data available, which can be used to train a supervised learning model.
  - There is also a large amount of unlabelled data available, which can be used to improve the performance of the supervised learning model.

- It is appropriate to utilise semi-supervised machine learning algorithms for this task.
  - In this case, the labelled data can be used to train a model, and the unlabelled data can be used to improve the model's performance.

- Inappropriate
