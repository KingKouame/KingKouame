```python
# Developing a Python script for sentiment analysis.

# First we load the spaCy English language model.

nlp = spacy.load("en_core_web_sm")

# Then we define a function to perform sentiment analysis on a single sentence

def analyze_sentiment(sentence):
  doc = nlp(sentence)
  sentiment = doc.sentiment
  return sentiment.score, sentiment.label
```

```python
# Preprocessing the text data.

reviews_data = df['reviews.text']
clean_data = reviews_data.dropna()

def clean_text(text):
  doc = nlp(text)
  cleaned_text = ' '.join([token.text for token in doc if not token.is_stop])
  return cleaned_text

clean_data['cleaned_text'] = clean_data.apply(clean_text)
```
