```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/balance.txt', sep=' ')
df.head(5)
```

```python
# This function will allow us to compare the average income based on ethnicity.

df.groupby('Ethnicity')['Income'].mean()
```

```python
# This function allow us to see the average balance between married and non-married people.

df.groupby('Married')['Balance'].mean()
```

```python
# This function allow us to see the hightest income in the dataset.

df['Income'].max()
```

```python
# This function allow us to see the lowest income in the dataset.

df['Income'].min()
```

```python
# This function allow us to see how many cards are recorded income in the dataset.

df['Cards'].sum()
```

```python
# This function allow us to see the number of females whom we have information for in comparison to males.

df['Gender'].value_counts()
```
