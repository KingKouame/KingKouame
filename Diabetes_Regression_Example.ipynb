```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/diabetes.csv')
df.head()
```

```python
# Differentiating between the independent variables and the dependent variable, and assigning them to variables X and Y.

# Independent variables (features)

X = df.iloc[:, 0:8]

# Dependent variable (target)

Y = df.iloc[:, 8]
```

```python
# Generating training and test sets comprising 80% and 20% of the data respectively.

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=0)
```

```python
# Using a MinMaxScaler and StandardScaler from sklearn.preprocessing, we will fit these scalers on the train set and use these fit scalers to transform the train and test sets.

from sklearn.preprocessing import MinMaxScaler, StandardScaler

# First we create the scalers.

scaler_minmax = MinMaxScaler()
scaler_standard = StandardScaler()

# Then, we fit the scalers on the train set.

scaler_minmax.fit(X_train)
scaler_standard.fit(X_train)

# Finally, we transform the train and test sets.

X_train_minmax = scaler_minmax.transform(X_train)
X_test_minmax = scaler_minmax.transform(X_test)

X_train_standard = scaler_standard.transform(X_train)
X_test_standard = scaler_standard.transform(X_test)
```

```python
# Generating a multiple linear regression model using the training set, using all of the independent variables.

# We first create a linear regression model.

model = LinearRegression()

# Then fit the model to the training data.

model.fit(X_train_minmax, y_train)

# Finally we print the print out the intercept and coefficients of the trained model.

print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
```

```python
# Generating predictions for the test set.

y_pred = model.predict(X_test_minmax)
print("Predicted values:", y_pred)
```

```python
# Computing R-squared for the model on the test set, using r2_score from sklearn.metrics to obtain this score.

from sklearn.metrics import r2_score

r2_score_model = r2_score(y_test, y_pred)
print("R-squared for the model on the test set:", r2_score_model)
```
